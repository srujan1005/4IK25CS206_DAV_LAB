```python
import pandas as pd
import numpy as np
```

```python
data={'Name':['Amit','Bebo','Chetan','Amit','Divya','Ravi','Sneha','Karan',
              'Pooja','Manish','Anjali','Suresh','Kavya','Rahul','Meera',
              'Vikram','Nisha','Arjun','Divya','Rohit','Swathi','Naveen',
              'Priya','Karan','Deepa'],
      'Age':[25,np.nan,30,25,28,22,24,np.nan,27,26,23,29,np.nan,
             25,22,28,24,np.nan,28,26,23,30,27,np.nan,25],
      'Marks':[85,90,np.nan,85,78,95,88,72,np.nan,91,67,84,79,
               np.nan,93,88,76,82,78,np.nan,90,85,71,72,89],
      'City':['Mysuru','Mysuru','Bangaluru','Mysuru',np.nan,'Delhi','Mysuru',
              'Mangalore','Bangaluru',np.nan,'Mysuru','Hubali','Mangalore',
              'Delhi','Bangaluru',np.nan,'Mysuru','Hubali',np.nan,'Delhi',
              'Mangalore','Mysuru','Bangaluru','Mangalore','Hubali']
     }
df = pd.DataFrame(data)
print("Dataset:")
print(df)
```

```python
print("\nShape of dataset (rows, columns):",df.shape)
```

```python
print("\nColumn names:",df.columns.tolist())
```

```python
print("\nData types of each column:")
print(df.dtypes)
```

```python
print("\nDataset Info:")
df.info()
```

```python
print("\nDescriptive statistics (numerical columns):")
print(df.describe())
```

```python
print("\nMean:\n",df.mean(numeric_only=True))
```

```python
print("\nMedian:\n",df.median(numeric_only=True))
```

```python
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values in dataset:",df.isnull().sum().sum())
```
